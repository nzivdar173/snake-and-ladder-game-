```python
import numpy as np

Game_board = np.arange(1, 101)[::-1]


def player_move(position):
    dice = np.random.randint(1, 7)
    print("Dice:", dice)

    if position + dice <= 100:
        position = position + dice

    return position

def ladder (position):
  if position == 4:
    position = 14
  elif position == 9:
    position = 31
  elif position == 28:
    position = 84
  elif position == 21:
    position = 42
  elif position == 51:
    position = 67
  elif position == 72:
    position = 91
  elif position == 80:
    position = 99

  return position


def snake (position):
  if position == 17:
    position = 7
  elif position == 54:
    position = 34
  elif position == 62:
    position = 19
  elif position == 64:
    position = 60
  elif position == 87:
    position = 36
  elif position == 92:
    position = 73
  elif position == 95:
    position = 75
  elif position == 98:
    position = 79

  return position


player_one_position = 0
player_two_position = 0

while player_one_position < 100 and player_two_position < 100:

    # Player 1
    input("Player 1, press Enter to roll the dice...")
    player_one_position = player_move(player_one_position)


    player_one_position = ladder(player_one_position)
    player_one_position = snake(player_one_position)

    print("Player 1 position:", player_one_position)

    if player_one_position == 100:
        print("Player 1 wins!")
        break

    if player_one_position == player_two_position:
      player_two_position = 0


    # Player 2
    input("Player 2, press Enter to roll the dice...")
    player_two_position = player_move(player_two_position)


    player_two_position = ladder(player_two_position)
    player_two_position = snake(player_two_position)

    print("Player 2 position:", player_two_position)

    if player_two_position == 100:
        print("Player 2 wins!")
        break

    if player_two_position == player_one_position:
      player_one_position = 0

```
